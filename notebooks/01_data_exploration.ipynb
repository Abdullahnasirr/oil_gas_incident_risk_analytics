```python
# Keep only the columns that matter for predicting significance
cols_to_keep = [
    'Significant',
    'What happened category',
    'Why it happened category',
    'Province',
    'Company',
    'Substance',
    'Release Type',
    'Pipeline or Facility Type',
    'Approximate Volume Released (m3)',
    'Year',
    'Latitude',
    'Longitude',
    'Emergency Level',
    'Population Density',
    'Land Use',
]

df_clean = df[cols_to_keep].copy()
print(df_clean.shape)
print(df_clean.head(3))
```

```python
# Fix 1: Convert Significant from Yes/No to 1/0
df_clean['Significant'] = df_clean['Significant'].map({'Yes': 1, 'No': 0})
print(df_clean['Significant'].value_counts())
```

```python
# Fix 2: Convert volume released to a proper number
df_clean['Approximate Volume Released (m3)'] = df_clean['Approximate Volume Released (m3)'].replace('Not Applicable', 0)
df_clean['Approximate Volume Released (m3)'] = pd.to_numeric(df_clean['Approximate Volume Released (m3)'], errors='coerce')
df_clean['Approximate Volume Released (m3)'] = df_clean['Approximate Volume Released (m3)'].fillna(0)

print(df_clean['Approximate Volume Released (m3)'].dtype)
print(df_clean['Approximate Volume Released (m3)'].describe())
```

```python
# Fix 3: Fill missing text columns with 'Unknown'
text_cols = ['What happened category', 'Why it happened category',
             'Pipeline or Facility Type', 'Emergency Level',
             'Population Density', 'Land Use']

df_clean[text_cols] = df_clean[text_cols].fillna('Unknown')

# Check how many nulls remain
print(df_clean.isnull().sum())
```

```python
# Save the cleaned dataset
df_clean.to_csv('../data/pipeline_incidents_clean.csv', index=False)
print("Saved!")
```
