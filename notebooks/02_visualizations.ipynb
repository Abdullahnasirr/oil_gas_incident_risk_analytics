```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/pipeline_incidents_clean.csv')
print(df.shape)
```

```python
# Chart 1: Incidents per year
plt.figure(figsize=(12, 5))
df.groupby('Year')['Incident Number'].count().plot(kind='bar', color='steelblue')
plt.title('Pipeline Incidents per Year (2008-2024)')
plt.xlabel('Year')
plt.ylabel('Number of Incidents')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('../dashboard/incidents_per_year.png', dpi=150)
plt.show()
```

```python
plt.figure(figsize=(12, 5))
df.groupby('Year').size().plot(kind='bar', color='steelblue')
plt.title('Pipeline Incidents per Year (2008-2024)')
plt.xlabel('Year')
plt.ylabel('Number of Incidents')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('../dashboard/incidents_per_year.png', dpi=150)
plt.show()
```

```python
# Chart 2: What caused incidents
plt.figure(figsize=(10, 6))
cause_counts = df['What happened category'].value_counts()
sns.barplot(x=cause_counts.values, y=cause_counts.index, palette='Blues_r')
plt.title('Incidents by Cause Category')
plt.xlabel('Number of Incidents')
plt.ylabel('')
plt.tight_layout()
plt.savefig('../dashboard/incidents_by_cause.png', dpi=150)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
cause_counts = df['What happened category'].value_counts().head(10)
sns.barplot(x=cause_counts.values, y=cause_counts.index,
            hue=cause_counts.index, palette='Blues_r', legend=False)
plt.title('Top 10 Incident Cause Categories')
plt.xlabel('Number of Incidents')
plt.ylabel('')
plt.tight_layout()
plt.savefig('../dashboard/incidents_by_cause.png', dpi=150)
plt.show()
```

```python
# Chart 3: Significant vs Non-Significant by cause
plt.figure(figsize=(10, 6))
sig_by_cause = df.groupby('What happened category')['Significant'].mean().sort_values(ascending=False).head(10)
sns.barplot(x=sig_by_cause.values, y=sig_by_cause.index,
            hue=sig_by_cause.index, palette='Reds_r', legend=False)
plt.title('Significance Rate by Cause Category')
plt.xlabel('Proportion of Incidents Classified as Significant')
plt.ylabel('')
plt.tight_layout()
plt.savefig('../dashboard/significance_by_cause.png', dpi=150)
plt.show()
```

```python
# Chart 4: Incidents by Province
plt.figure(figsize=(10, 5))
province_counts = df['Province'].value_counts()
sns.barplot(x=province_counts.index, y=province_counts.values,
            hue=province_counts.index, palette='Blues_r', legend=False)
plt.title('Pipeline Incidents by Province')
plt.xlabel('Province')
plt.ylabel('Number of Incidents')
plt.tight_layout()
plt.savefig('../dashboard/incidents_by_province.png', dpi=150)
plt.show()
```
